# 00c · Open the sealed answer keys (grading time)

**Only run this after** the frozen model (MMM v2) and the frozen budget recommendation are on GitHub. Their commit dates prove every decision was made before the answers were seen.

This notebook:
1. Reads `truth.json` and `experiment_truth.json` from your Drive folder `prakriti_sealed`.
2. Checks each file's fingerprint against the one committed on GitHub (21 Oct). If they match, nobody changed the answers.
3. Downloads both files so they can be published and used for grading.

```
REPO_URL = 'https://github.com/anvesha2402/prakriti-growth-measurement.git'   # change only if your repo name is different
!rm -rf /content/repo
!git clone -q $REPO_URL /content/repo
%cd /content/repo
import os, hashlib, shutil
assert os.path.exists('model/mmm_v2_FROZEN.json'), 'Upload the frozen MMM v2 files to GitHub first.'
assert os.path.exists('model/optimiser_FROZEN.txt'), 'Upload the frozen optimiser files to GitHub first.'
print('Frozen model and recommendation found on GitHub.')
```

```
from google.colab import drive
drive.mount('/content/drive')
SEALED = '/content/drive/MyDrive/prakriti_sealed'
checks = [('truth.json', 'sealed/HASH.txt'), ('experiment_truth.json', 'sealed/HASH_stage2.txt')]
for fname, hash_file in checks:
    committed = open(hash_file).read().split(':')[1].split()[0].strip()
    current = hashlib.sha256(open(f'{SEALED}/{fname}', 'rb').read()).hexdigest()
    print(f'{fname:24s}', 'MATCH' if current == committed else 'MISMATCH', current[:16], '...')
    assert current == committed, f'{fname} changed since it was sealed: stop and report this.'
    shutil.copy(f'{SEALED}/{fname}', f'sealed/{fname}')
print('Both answer keys are unchanged since sealing.')
```
