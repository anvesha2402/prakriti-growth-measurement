## Part 3 · The SEALED run (run once)
The computer picks a secret random number nobody sees, builds the company, saves weeks 1–87 for you to study, and locks the answers in your Drive.

You will see only a **fingerprint** (a long code). Copy it; it goes on GitHub.

```
stage1, fingerprint = pw.run_stage1(out_dir='data/generated/stage1', sealed_dir=SEALED)
import shutil, os
os.makedirs('sealed', exist_ok=True)
shutil.copy(f'{SEALED}/HASH.txt', 'sealed/HASH.txt')   # only the fingerprint goes to GitHub
print(open('sealed/HASH.txt').read())
```
